````python
documents = {
    'doc1': "The quick brown fox jumps over the lazy dog. The dog is lazy.",
    'doc2': "A brown cat sat on the mat. The cat is quick.",
    'doc3': "Foxes are clever. Dogs are loyal. Cats are independent.",
    'doc4': "Quick dogs and lazy cats are good pets."
}

print("Sample Documents:")
for doc_id, content in documents.items():
    print(f"{doc_id}: {content}")
````

````python
import re
import nltk
from nltk.corpus import stopwords

# Download NLTK stop words if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # Tokenize and convert to lowercase
    tokens = re.findall(r'\b\w+\b', text.lower())
    # Remove stop words
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

processed_documents = {}
for doc_id, content in documents.items():
    processed_documents[doc_id] = preprocess_text(content)

print("Processed Documents:")
for doc_id, tokens in processed_documents.items():
    print(f"{doc_id}: {tokens}")
````

### Step 3: Build the Inverted Index

An inverted index maps each unique term to a list of documents in which it appears. For our basic implementation, we'll store a dictionary where keys are terms and values are sets of document IDs.

Using sets ensures that each document ID is stored only once for a given term, and it provides efficient lookup and union operations for query processing.

````python
inverted_index = {}

for doc_id, tokens in processed_documents.items():
    for token in tokens:
        if token not in inverted_index:
            inverted_index[token] = set()
        inverted_index[token].add(doc_id)

print("Inverted Index:")
for term, doc_ids in inverted_index.items():
    print(f"'{term}': {sorted(list(doc_ids))}")
````

````python
def search_inverted_index(query, inverted_index, documents):
    processed_query = preprocess_text(query)
    print(f"Processed query: {processed_query}")

    if not processed_query:
        return []

    # Start with the document set for the first term
    # If the term is not in the index, start with an empty set
    results = inverted_index.get(processed_query[0], set()).copy()

    # Intersect with results for subsequent terms (AND logic)
    for term in processed_query[1:]:
        if not results: # Optimization: if results are already empty, no need to check further
            break
        term_docs = inverted_index.get(term, set())
        results = results.intersection(term_docs)

    if not results:
        return []
    else:
        # Return the actual document content for found doc_ids
        found_docs_content = {doc_id: documents[doc_id] for doc_id in sorted(list(results))}
        return found_docs_content

# Test the search function
query1 = "quick brown fox"
retrieved_docs1 = search_inverted_index(query1, inverted_index, documents)
print(f"\nDocuments for query '{query1}':")
if retrieved_docs1:
    for doc_id, content in retrieved_docs1.items():
        print(f"  {doc_id}: {content}")
else:
    print("  No documents found.")

query2 = "lazy cats"
retrieved_docs2 = search_inverted_index(query2, inverted_index, documents)
print(f"\nDocuments for query '{query2}':")
if retrieved_docs2:
    for doc_id, content in retrieved_docs2.items():
        print(f"  {doc_id}: {content}")
else:
    print("  No documents found.")

query3 = "elephant"
retrieved_docs3 = search_inverted_index(query3, inverted_index, documents)
print(f"\nDocuments for query '{query3}':")
if retrieved_docs3:
    for doc_id, content in retrieved_docs3.items():
        print(f"  {doc_id}: {content}")
else:
    print("  No documents found.")
````

````python
# 📝 Text Preprocessing: Stop Words, Stemming, Lemmatization (मराठीत)

## भाग 1: Basic Terms

| Term | सोप्या भाषेत अर्थ |
|---|---|
| **Text Preprocessing** | Raw text ला analysis साठी स्वच्छ आणि एकसारखं करणे. IR (Information Retrieval) चं पहिलं पाऊल |
| **NLTK** | Natural Language Toolkit. Python ची text processing library |
| **Token** | text चा लहान तुकडा (शब्द किंवा चिन्ह). उदा. `sentence`, `,` |
| **Tokenization** | वाक्याचे tokens मध्ये तुकडे करणे |
| **Stop Words** | खूप सामान्य शब्द ज्यांना स्वतःचा विशेष अर्थ नसतो (the, is, an, of, and). Search मध्ये उपयोग नाही |
| **Stemming** | शब्दाचे प्रत्यय (suffix) कापून मूळ रूप (stem) मिळवणे. Rule-based आणि अंदाजे. Stem खरा शब्द असेलच असं नाही |
| **Lemmatization** | शब्दाचं dictionary मधलं खरं मूळ रूप (lemma) शोधणे. व्याकरण आणि WordNet dictionary वापरतो |
| **Porter Stemmer** | सर्वात प्रसिद्ध stemming algorithm (Martin Porter, 1980) |
| **WordNet** | इंग्रजी शब्दांचा मोठा dictionary/database, lemmatizer हा वापरतो |
| **POS (Part of Speech)** | शब्दाचा प्रकार: noun (n), verb (v), adjective (a), adverb (r) |
| **Corpus** | text documents चा संग्रह |

**Stemming vs Lemmatization (उदाहरण):**

| शब्द | Stemming | Lemmatization (verb) |
|---|---|---|
| running | run | run |
| studies | studi ❌ | study ✅ |
| better | better | good (pos='a' असताना) |
| removal | remov ❌ | removal |

**पूर्ण flow:** Original text → Tokenization → Stop word removal → Stemming / Lemmatization

---

## भाग 2: Line-by-Line Explanation

### Imports

```python
import nltk
```
मुख्य NLTK library.

```python
from nltk.corpus import stopwords
```
इंग्रजी stop words ची ready list.

```python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```
Stemming आणि lemmatization साठी classes.

```python
from nltk.tokenize import word_tokenize
```
वाक्याला शब्दांत तोडणारं function.

### Resources Download

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```
NLTK मध्ये data वेगळा download करावा लागतो:
- `stopwords`: stop words ची list.
- `punkt` आणि `punkt_tab`: tokenizer चे models. नवीन NLTK versions ला `punkt_tab` लागतो.
- `wordnet`: lemmatizer चा dictionary.
- `omw-1.4`: Open Multilingual WordNet, WordNet सोबत लागतो.

### Function 1: Stop Word Removal

```python
def remove_stopwords(text):
```
`text` (string) घेणारं function.

```python
    stop_words = set(stopwords.words('english'))
```
इंग्रजी stop words ची list घेऊन **set** मध्ये बदलतो. **का set?** Set मध्ये शोध (`in`) खूप वेगवान असतो, list मध्ये हळू.

```python
    word_tokens = word_tokenize(text)
```
Text चे tokens बनवतो. उदा. `"Hello, world."` → `['Hello', ',', 'world', '.']`. साध्या `split()` पेक्षा चांगलं, कारण चिन्हं वेगळी होतात.

```python
    filtered_text = [
        w for w in word_tokens
        if w.lower() not in stop_words
    ]
```
**List comprehension:** प्रत्येक token `w` साठी, जर `w.lower()` stop words मध्ये **नसेल** तर ठेवतो. `.lower()` कारण stop word list लहान अक्षरात आहे (`This` ला `this` करून तपासावं लागतं). पण शब्द मूळ case मध्येच ठेवतो.

```python
    return " ".join(filtered_text)
```
Tokens ना स्पेसने जोडून पुन्हा string बनवतो.

### Function 2: Stemming

```python
def apply_stemming(text):
    ps = PorterStemmer()
    word_tokens = word_tokenize(text)
    stemmed_text = [ps.stem(w) for w in word_tokens]
    return " ".join(stemmed_text)
```
- `PorterStemmer()` चा object बनवतो.
- Tokens बनवतो.
- प्रत्येक token वर `ps.stem(w)` लावतो. उदा. `running → run`, `learning → learn`.
- Porter stemmer आपोआप lowercase करतो.
- पुन्हा string करून परत करतो.

### Function 3: Lemmatization

```python
def apply_lemmatization(text):
    lemmatizer = WordNetLemmatizer()
    word_tokens = word_tokenize(text)
    lemmatized_text = [
        lemmatizer.lemmatize(w.lower(), pos='v')
        for w in word_tokens
    ]
    return " ".join(lemmatized_text)
```
- `WordNetLemmatizer()` चा object.
- `w.lower()`: WordNet मध्ये शब्द lowercase असतात.
- **`pos='v'`**: प्रत्येक शब्द **verb** समजून lemmatize करतो. Default `pos='n'` (noun) असतो. म्हणून `running` आणि `learning` verb म्हणून `run` आणि `learn` होतात. Noun म्हणून तसेच राहिले असते.
- **मर्यादा:** सर्व शब्द verb समजले जातात, म्हणून noun जसे `things` बदलत नाहीत.

### Main Program

```python
example_text = "This is an example sentence, ... Running and learning new things are great!"
print(f"Original Text: {example_text}\n")
```
नमुना वाक्य आणि ते छापणे. `\n` = नवीन ओळ.

```python
text_without_stopwords = remove_stopwords(example_text)
stemmed_text = apply_stemming(text_without_stopwords)
lemmatized_text = apply_lemmatization(text_without_stopwords)
```
- आधी stop words काढतो.
- मग **stop words काढलेल्या** text वर stemming आणि lemmatization **स्वतंत्रपणे** लावतो. Lemmatization हे stemmed text वर नाही, दोन्ही समांतर आहेत.

### अपेक्षित Output (मी हाताने काढलेलं, Colab मधील output शी तपासा)

**After Stop Word Removal:**
```
example sentence , showing stop word removal stemming functionalities NLTK . Running learning new things great !
```
(काढलेले: This, is, an, the, off, and, of, are)

**After Stemming:**
```
exampl sentenc , show stop word remov stem function nltk . run learn new thing great !
```

**After Lemmatization:**
```
example sentence , show stop word removal stem functionalities nltk . run learn new things great !
```

**निरीक्षण:** Stemming ने `exampl`, `sentenc`, `remov` असे खरे नसलेले शब्द दिले. Lemmatization ने खरे शब्द ठेवले (`example`, `sentence`, `removal`).

---

## भाग 3: Viva Q&A

**Q1. Text preprocessing म्हणजे काय आणि का आवश्यक?**
उत्तर: Raw text ला स्वच्छ आणि प्रमाणित करणे. यामुळे noise कमी होतो, index लहान होतो आणि search/retrieval ची अचूकता वाढते.

**Q2. Tokenization म्हणजे काय?**
उत्तर: Text ला शब्द किंवा चिन्हांच्या लहान तुकड्यांत (tokens) तोडणे.

**Q3. Stop words म्हणजे काय? ते का काढतात?**
उत्तर: the, is, an, of सारखे अतिसामान्य शब्द ज्यांचा अर्थात फारसा वाटा नसतो. ते काढल्याने index चा आकार कमी होतो आणि प्रक्रिया वेगवान होते.

**Q4. Stop words काढण्याचे तोटे?**
उत्तर: काही वेळा अर्थ बदलतो. उदा. "to be or not to be", "not" काढला तर अर्थ उलटा होतो. Phrase queries मध्येही अडचण येते.

**Q5. Stemming म्हणजे काय?**
उत्तर: शब्दाचे suffix कापून stem मिळवणे. उदा. running → run. हे rule-based आणि वेगवान आहे, पण stem खरा शब्द असेलच असं नाही (studies → studi).

**Q6. Lemmatization म्हणजे काय?**
उत्तर: शब्दाचा dictionary मधला खरा मूळ रूप (lemma) शोधणे, व्याकरण आणि POS विचारात घेऊन. उदा. better → good (adjective म्हणून).

**Q7. Stemming आणि Lemmatization मध्ये फरक?**

| Stemming | Lemmatization |
|---|---|
| Suffix कापतो (rule-based) | Dictionary (WordNet) वापरतो |
| वेगवान | हळू |
| Output खरा शब्द नसू शकतो | Output नेहमी खरा शब्द |
| POS लागत नाही | POS चा परिणाम होतो |
| कमी अचूक | जास्त अचूक |

**Q8. Over-stemming आणि Under-stemming म्हणजे काय?**
उत्तर: **Over-stemming:** वेगळे अर्थाचे शब्द एकाच stem वर येतात (university, universe → univers). **Under-stemming:** एकाच मुळाचे शब्द वेगळे stems देतात (data, datum).

**Q9. Porter Stemmer म्हणजे काय?**
उत्तर: Martin Porter चा 1980 मधला rule-based algorithm जो टप्प्याटप्प्याने (5 steps) suffix काढतो. इंग्रजीसाठी सर्वाधिक वापरला जातो.

**Q10. इतर stemmers कोणते?**
उत्तर: Lancaster (आक्रमक, जास्त कापतो), Snowball/Porter2 (सुधारित आणि बहुभाषिक).

**Q11. या code मध्ये `pos='v'` का दिलंय? ते नसतं तर?**
उत्तर: WordNet ला शब्दाचा प्रकार सांगावा लागतो. `pos='v'` दिल्याने `running → run`, `learning → learn` होतात. Default noun असतं तर हे शब्द तसेच राहिले असते.

**Q12. `pos='v'` ची मर्यादा काय?**
उत्तर: सर्व शब्द verb मानले जातात, म्हणून noun/adjective योग्य बदलत नाहीत (उदा. `things` तसाच राहतो). योग्य उपाय: आधी POS tagging करून प्रत्येक शब्दाला योग्य pos देणे.

**Q13. `stopwords.words('english')` ला `set()` का लावलं?**
उत्तर: Set मध्ये lookup O(1) वेळेत होतो, list मध्ये O(n). Text मोठा असेल तर फरक पडतो.

**Q14. `w.lower()` का वापरलं?**
उत्तर: Stop word list lowercase आहे. `This` ला `this` करून तपासलं नाही तर तो काढला जाणार नाही.

**Q15. `word_tokenize` आणि `split()` मध्ये फरक?**
उत्तर: `split()` फक्त स्पेसवर तोडतो, म्हणून `sentence,` असं चिन्हासहित येतं. `word_tokenize` चिन्हं वेगळे tokens करतं आणि contractions (don't → do, n't) नीट हाताळतं.

**Q16. `punkt` म्हणजे काय?**
उत्तर: Sentence/word tokenization साठी pre-trained model.

**Q17. IR मध्ये preprocessing चा क्रम काय असतो?**
उत्तर: Tokenization → Lowercasing → Stop word removal → Stemming/Lemmatization → Indexing.

**Q18. Preprocessing चे उपयोग कुठे होतात?**
उत्तर: Search engines, sentiment analysis, spam filtering, document classification, chatbots, topic modeling.

**Q19. `' '.join()` काय करतं?**
उत्तर: Tokens च्या list ला स्पेस टाकून एका string मध्ये जोडतं.

**Q20. Stemming आणि Lemmatization पैकी IR साठी काय चांगलं?**
उत्तर: वेग आणि recall महत्त्वाचा असेल तर stemming. अचूकता आणि खरे शब्द हवे असतील तर lemmatization.

---

## ⚠️ Viva साठी महत्त्वाच्या टिपा

1. **Stemming आणि lemmatization दोन्ही stop words काढलेल्या text वर वेगळे वेगळे चालतात.** एकाचा output दुसऱ्याला दिला जात नाही.
2. **Punctuation (`,` `.` `!`) काढले जात नाहीत.** ते token म्हणून राहतात, कारण stop word list मध्ये चिन्हं नाहीत. Examiner ने विचारलं तर सांगा की यासाठी `string.punctuation` किंवा `isalpha()` वापरून फिल्टर करता येतं.
3. **`NLTK` stemming मध्ये `nltk` (lowercase) होतं**, पण stop word removal मध्ये case तसाच राहतो.
4. **`new` stop word नाही**, म्हणून तो राहतो. `off` stop word आहे, म्हणून तो काढला जातो.
5. `nltk.download(...)` internet लागतो आणि फक्त एकदा करावा लागतो.

All the best for the practical! 🙂 पुढचा code आला की पाठव.
````
